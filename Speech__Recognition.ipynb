```python
import speech_recognition as sr
import pyttsx3
import os

!apt-get update
!apt-get install -y espeak-ng

!apt-get install -y portaudio19-dev
!pip install pyaudio
```

```python

recognizer = sr.Recognizer()

engine = pyttsx3.init()

audio_fike_path = 'Shubham_.wav'

# if file not found
if not os.path.exists(audio_fike_path):
    print(f"Audio file '{audio_fike_path}' not found.")
else:
  # source is the open audio file
    with sr.AudioFile(audio_fike_path) as source:
        print(f"Processing audio from '{audio_fike_path}'")
        audio = recognizer.record(source) # -> Read the audio data from source

    try:
      print("Recognizing...")
      text = recognizer.recognize_google(audio)
      print(f"You said: {text}")

      # Convert text to speech
      engine.say(text)
      # speak the text
      engine.runAndWait()


# if system cannot understand the audio data
    except sr.UnknownValueError:
      print("Sorry, I couldn't understand the audio.")
    except sr.RequestError as e:
      print(f"Could not request results from Google Speech Recognition")
```

```python
import pyttsx3

engine = pyttsx3.init()

text = """
  Hello Shubham,
  welcome to the Artifical Intelligence laboratory.
  This is a sample voice recording for the speech to text and text to speech conversion.
  Have a good day.
"""

engine.setProperty('rate', 150)
engine.setProperty('volume', 1.0)

engine.save_to_file(text, 'hello_shubham.wav')
engine.runAndWait()

print("Wav file created successfully")
```
